```python
df['Review Rating'] = (
    df.groupby('Category')['Review Rating']
      .transform(lambda x: x.fillna(x.median()))
)
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')

df = df.rename(
    columns={'purchase_amount_(usd)': 'purchase_amount'}
)
```

```python
bins = [0, 18, 40, 60, float('inf')]

labels = [
    'Young',
    'Adult',
    'Middle Aged',
    'Senior Citizen'
]

df['age_group'] = pd.cut(
    df['age'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
frequency_mapping = {
    'Weekly': 7,
    'Fortnightly': 14,
    'Bi-Weekly': 14,
    'Monthly': 30,
    'Every 3 Months': 90,
    'Quarterly': 90,
    'Annually': 365
}

df['purchase_frequency_days'] = (
    df['frequency_of_purchases'].map(frequency_mapping)
)
```

```python
from sqlalchemy import create_engine
```

```python
username = "postgres"
password = 0000
host = "localhost"
port = "5432"
database = "customer_behavior"

engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

print("PostgreSQL engine created.")
```

```python
from sqlalchemy import text

with engine.connect() as connection:
    result = connection.execute(text("SELECT current_database();"))
    print("Connected to:", result.scalar())
```

```python
table_name = "customer"

df.to_sql(
    table_name,
    engine,
    if_exists="replace",
    index=False
)

print(
    f"Data successfully loaded into table '{table_name}' "
    f"in database '{database}'."
)
```

```python
query = """
SELECT
    gender,
    ROUND(AVG(purchase_amount), 2) AS avg_purchase_amount
FROM customer
GROUP BY gender
ORDER BY avg_purchase_amount DESC;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT
    age_group,
    COUNT(*) AS total_customers,
    SUM(purchase_amount) AS total_revenue
FROM customer
GROUP BY age_group
ORDER BY total_revenue DESC;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT
    item_purchased,
    COUNT(*) AS total_purchases,
    SUM(purchase_amount) AS total_revenue
FROM customer
GROUP BY item_purchased
ORDER BY total_revenue DESC
LIMIT 10;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT
    subscription_status,
    COUNT(*) AS total_customers,
    ROUND(AVG(purchase_amount), 2) AS avg_purchase_amount,
    SUM(purchase_amount) AS total_revenue
FROM customer
GROUP BY subscription_status
ORDER BY total_revenue DESC;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT
    discount_applied,
    COUNT(*) AS total_purchases,
    ROUND(AVG(purchase_amount), 2) AS avg_purchase_amount,
    SUM(purchase_amount) AS total_revenue
FROM customer
GROUP BY discount_applied
ORDER BY total_revenue DESC;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT
    frequency_of_purchases,
    purchase_frequency_days,
    COUNT(*) AS total_customers,
    ROUND(AVG(purchase_amount), 2) AS avg_purchase_amount
FROM customer
GROUP BY frequency_of_purchases, purchase_frequency_days
ORDER BY purchase_frequency_days;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT
    payment_method,
    COUNT(*) AS total_transactions,
    SUM(purchase_amount) AS total_revenue,
    ROUND(AVG(purchase_amount), 2) AS avg_purchase_amount
FROM customer
GROUP BY payment_method
ORDER BY total_revenue DESC;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT
    location,
    COUNT(*) AS total_purchases,
    SUM(purchase_amount) AS total_revenue
FROM customer
GROUP BY location
ORDER BY total_revenue DESC
LIMIT 10;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT
    customer_id,
    previous_purchases,
    purchase_amount,
    CASE
        WHEN previous_purchases >= 40 THEN 'Loyal Customer'
        WHEN previous_purchases >= 20 THEN 'Regular Customer'
        WHEN previous_purchases >= 10 THEN 'Occasional Customer'
        ELSE 'New Customer'
    END AS customer_segment
FROM customer
ORDER BY previous_purchases DESC;
"""

segments = pd.read_sql(query, engine)

segments.head(10)
```

```python
query = """
SELECT
    CASE
        WHEN previous_purchases >= 40 THEN 'Loyal Customer'
        WHEN previous_purchases >= 20 THEN 'Regular Customer'
        WHEN previous_purchases >= 10 THEN 'Occasional Customer'
        ELSE 'New Customer'
    END AS customer_segment,

    COUNT(*) AS total_customers,
    ROUND(AVG(purchase_amount), 2) AS avg_purchase_amount,
    SUM(purchase_amount) AS total_revenue

FROM customer

GROUP BY customer_segment
ORDER BY total_revenue DESC;
"""

pd.read_sql(query, engine)
```

```python
from sqlalchemy import text

query = """
ALTER TABLE customer
ADD COLUMN IF NOT EXISTS customer_segment VARCHAR(30);
"""

with engine.begin() as connection:
    connection.execute(text(query))

print("customer_segment column created.")
```

```python
query = """
UPDATE customer
SET customer_segment =
    CASE
        WHEN previous_purchases >= 40 THEN 'Loyal Customer'
        WHEN previous_purchases >= 20 THEN 'Regular Customer'
        WHEN previous_purchases >= 10 THEN 'Occasional Customer'
        ELSE 'New Customer'
    END;
"""

with engine.begin() as connection:
    connection.execute(text(query))

print("Customer segmentation completed.")
```
